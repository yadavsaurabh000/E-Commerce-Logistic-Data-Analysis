```python
import pandas as pd
import numpy as np
```

```python
df =pd.read_csv('retail-orders-raw.csv')
```

```python
df.shape
```

```python
df
```

```python

duplicate_count = df.duplicated().sum()
# print(f"Total duplicate rows: {duplicate_count}")

```

```python
# Duplicate rows remove karein
df = df.drop_duplicates()
print("Duplicates hatane ke baad rows:", len(df))
```

```python
df
```

```python
df['order_date'] = df['order_date'].fillna('2026-01-01')
```

```python
df['customer_segment']=df['customer_segment'].str.strip().str.title()
```

```python
df=df.reset_index(drop=True)
```

```python
df
```

```python
df['city']=df['city'].fillna('Unknown')
```

```python
df
```

```python


df['order_date']=pd.to_datetime(df['order_date'], errors='coerce',format='mixed')
```

```python
df.loc[6 ,'order_date']=pd.to_datetime('2026-01-12')
```

```python
df
```

```python
df.loc[5,'order_date']=pd.to_datetime('2026-10-13')
```

```python
df['quantity']=df['quantity'].replace({'two':2, -1 : 1, '-1':1})

df['quantity'] = pd.to_numeric(df['quantity'])
```

```python
df
```

```python
df['discount_pct']=df['discount_pct'].fillna(0.0)
```

```python
df['discount_pct']=df['discount_pct'].replace(105.0 , 10.5)
```

```python
df
```

```python
df['payment_status']=df['payment_status'].str.title()
```

```python
df
```

```python
total_quantity = df['quantity'].sum()
print(f"1. Total quantity: {total_quantity}")
```

```python
total_revenue = df['unit_price'].sum()
print(f"2. Total Revenue: ₹{total_revenue:.2f}")
```

```python
total_orders = df['order_id'].nunique()
print(f"3.  Total Orders  : {total_orders}")
```

```python
paid_pct = (df['payment_status'] == 'Paid').mean() * 100
print(f"4. Successful Payment Rate : {paid_pct:.2f}%")
```

```python
avg_discount = df['discount_pct'].mean()
print(f"5. Average Discount Percentage  : {avg_discount:.2f}%")
```

```python
average_price = df['unit_price'].mean()

print(f"6. Average Unit Price: ₹{average_price:.2f}")
```

```python
top_city = df['city'].value_counts().idxmax()
print(f"7. Top City by Sales : {top_city}")
```

```python
for seg in df['customer_segment'].unique():
    # Har segment ka data filter karein
    sub_df = df[df['customer_segment'] == seg]
    
    seg_orders = sub_df['order_id'].count()
    seg_qty = sub_df['quantity'].sum()
    seg_revenue = sub_df['unit_price'].sum()
    
    print(f"   • Segment: {seg}")
    print(f"     - Orders   : {seg_orders}")
    print(f"     - Quantity : {seg_qty}")
    print(f"     - Revenue  : ₹{seg_revenue:.2f}\n")
```
